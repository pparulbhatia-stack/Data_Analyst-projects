```python
import pandas as pd
```

```python
data = pd.read_csv("california_housing_train.csv")
```

```python
print("Dataset size:", data.shape)

print("\nColumn names:")
print(data.columns)

print("\nData types:")
print(data.dtypes)

print("\nMissing values:")
print(data.isnull().sum())
```

```python
#dataset summary

print(data.describe())

#duplicate records (average median house value)

print("Duplicate rows:", data.duplicated().sum())

#Task 1: Income & house price relation

correlation = data["median_income"].corr (data["median_house_value"])
print("correlation between median income and median house values:", correlation)

#Income groups average median house value kitni hai.

data["income_group"] = pd.cut( 
    data["median_income"],
    bins=4,
    labels=["Low", "Medium-Low", "Medium-High", "High"]
)

result = data.groupby(
    "income_group",
    observed=True
)["median_house_value"].mean()

print(result)




```

```python
import matplotlib.pyplot as plt

print("Matplotlib installed successfully!")
```

```python
result.plot(kind="bar")

plt.title("Average House Value by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Average Median House Value ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


#House values ka distribution 

plt.figure(figsize=(8, 5))

plt.hist(data["median_house_value"], bins=30)

plt.title("Distribution of Median House Values")
plt.xlabel("Median House Value ($)")
plt.ylabel("Number of Records")

plt.tight_layout()
plt.show()

#California  geographical analysis


plt.figure(figsize=(8, 5))

plt.scatter(
    data["longitude"],
    data["latitude"],
    c=data["median_house_value"],
    alpha=0.4
)

plt.title("California Housing Prices by Location")
plt.xlabel("Longitude")
plt.ylabel("Latitude")

plt.colorbar(label="Median House Value ($)")
plt.tight_layout()
plt.show()

```
